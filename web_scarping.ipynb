```python
from bs4 import BeautifulSoup
import os
```

```python
def load_html_file(file_path):
    """Load the local HTML file"""
    
    if not os.path.exists(file_path):
        print(f"Error: File '{file_path}' not found.")
        return None
    
    with open(file_path, 'r', encoding='utf-8') as file:
        soup = BeautifulSoup(file, 'html.parser')
    
    print("HTML file loaded successfully!")
    return soup
```

```python
def scrape_training_programs(soup):
    """Scrape training program information"""
    print("\n=== Training Programs ===")
    programs = []
    
    program_cards = soup.select('.program-cards .program-card')
    
    for card in program_cards:
        program = {
            'title': card.select_one('h3').text.strip(),
            'duration': card.select('.program-meta span')[0].text.strip(),
            'format': card.select('.program-meta span')[1].text.strip(),
            'description': card.select('p')[0].text.strip(),
            'start_date': card.select('.program-meta')[1].select('span')[0].text.replace('Next Cohort: ', '').strip(),
            'price': card.select('.program-meta')[1].select('span')[1].text.strip()
        }
        programs.append(program)
    
    print(f"Found {len(programs)} training programs:")
    return programs
```

```python
training_programs = scrape_training_programs(response)
training_programs
```

```python
def scrape_upcoming_events(soup):
    """Scrape upcoming events from the table"""
    print("\n=== Upcoming Events ===")
    events = []
    
    events_table = soup.find('table')
    rows = events_table.find('tbody').find_all('tr')
    
    for row in rows:
        cols = row.find_all('td')
        event = {
            'date': cols[0].text.strip(),
            'name': cols[1].text.strip(),
            'location': cols[2].text.strip(),
            'type': cols[3].text.strip(),
            'price': cols[4].text.strip(),
            'highlighted': 'event-highlight' in row.get('class', [])
        }
        events.append(event)
    
    print(f"Found {len(events)} upcoming events:")
    return events
```

```python
upcoming_events = scrape_upcoming_events(response)
upcoming_events
```

```python
def scrape_latest_articles(soup):
    """Scrape the latest articles"""
    print("\n=== Latest Articles ===")
    articles = []
    
    article_cards = soup.select('.article-cards .article-card')
    
    for card in article_cards:
        article = {
            'title': card.select_one('h3').text.strip(),
            'date': card.select('.article-meta span')[0].text.strip(),
            'read_time': card.select('.article-meta span')[2].text.strip(),
            'description': card.select('p')[0].text.strip(),
            'tags': [tag.text.strip() for tag in card.select('.article-tag')]
        }
        articles.append(article)
    
    print(f"Found {len(articles)} articles:")
    return articles
```

```python
latest_articles = scrape_latest_articles(response)
latest_articles
```

```python
def scrape_membership_plans(soup):
    """Scrape membership plans and benefits"""
    print("\n=== Membership Plans ===")
    plans = []
    
    benefit_cards = soup.select('.benefit-card')
    
    for card in benefit_cards:
        plan = {
            'name': card.select_one('h3').text.strip(),
            'price': card.select('p')[0].text.strip(),
            'benefits': [li.text.strip() for li in card.select('li')]
        }
        plans.append(plan)
    
    print("Found membership plans:")
    return plans
```

```python
membership_plans = scrape_membership_plans(response)
membership_plans
```

```python
def scrape_contact_info(soup):
    """Scrape contact information from footer"""
    print("\n=== Contact Information ===")
    
    footer_links = [a.text.strip() for a in soup.select('.footer-links a')]
    social_links = [a.text.strip() for a in soup.select('.social-links a')]
    copyright_text = soup.select_one('.copyright').text.strip()
    
    contact_info = {
        'footer_links': footer_links,
        'social_media': social_links,
        'copyright': copyright_text
    }
    
    print("Contact and footer information:")
    return contact_info
```

```python
contact_info = scrape_contact_info(response)
contact_info
```
